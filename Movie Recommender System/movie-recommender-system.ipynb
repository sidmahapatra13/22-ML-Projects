```python
df_movies_merged = df_movies.merge(df_credits, on='title')
df_movies_merged.sample(5)
```

```python
# delete unnecessary cols.

df_movies_merged = df_movies_merged[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
```

```python
import ast

def convert(obj):
    # If it's already a list, use it directly; otherwise parse the string
    if isinstance(obj, str):
        try:
            obj = ast.literal_eval(obj)
        except (ValueError, SyntaxError):
            return []

    if not isinstance(obj, list):
        return []

    # Handle both list-of-dicts (TMDB style) and list-of-strings
    return [i['name'] if isinstance(i, dict) else i for i in obj]
```

```python
df_movies_merged['genres'] = df_movies_merged['genres'].apply(convert)
df_movies_merged['keywords'] = df_movies_merged['keywords'].apply(convert)
```

```python
def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter+=1
        else:
            break
    return L
```

```python
df_movies_merged['cast'] = df_movies_merged['cast'].apply(convert3)
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L
```

```python
df_movies_merged['crew'] = df_movies_merged['crew'].apply(
	lambda x: fetch_director(x) if isinstance(x, str) else x
)
df_movies_merged['crew']
```

```python
df_movies_merged.sample(5)
```

```python
df_movies_merged['overview'][0]
```

```python
df_movies_merged['overview'] = df_movies_merged['overview'].apply(lambda x:x.split())
df_movies_merged['overview']
```

```python
#remove all spaces so that names/genres containing 2 words are not treated as different tags

df_movies_merged['genres'] = df_movies_merged['genres'].apply(lambda x:[i.replace(" ", "") for i in x])
df_movies_merged['keywords'] = df_movies_merged['keywords'].apply(lambda x:[i.replace(" ", "") for i in x])
df_movies_merged['cast'] = df_movies_merged['cast'].apply(lambda x:[i.replace(" ", "") for i in x])
df_movies_merged['crew'] = df_movies_merged['crew'].apply(lambda x:[i.replace(" ", "") for i in x])
```

```python
df_movies_merged.sample(3)
```

```python
#combine all relevant columns into 1 "tags" column

df_movies_merged['tags'] = df_movies_merged['overview'] + df_movies_merged['genres'] + df_movies_merged['keywords'] + df_movies_merged['cast'] + df_movies_merged['crew']
new_df = df_movies_merged[['movie_id', 'title', 'tags']]

new_df
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
new_df['tags']
```

```python
new_df['tags'][3]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())
```

**Text Stemming**

Ex: Loved, Loving, Love -> Love

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))
        
    return " ".join(y)
```

```python
stem("following the death of district attorney harvey dent, batman assumes responsibility for dent's crimes to protect the late attorney's reputation and is subsequently hunted by the gotham city police department. eight years later, batman encounters the mysterious selina kyle and the villainous bane, a new terrorist leader who overwhelms gotham's finest. the dark knight resurfaces to protect a city that has branded him an enemy. action crime drama thriller dccomics crimefighter terrorist secretidentity burglar hostagedrama timebomb gothamcity vigilante cover-up superhero villainess tragichero terrorism destruction catwoman catburglar imax flood criminalunderworld batman christianbale michaelcaine garyoldman christophernolan")
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

<h2>Vectorization</h2>
<hr>

**Bag of Words Method**

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
vectors
```

```python
#identifying the 5000 most freq. words

list(cv.get_feature_names_out())
```

**Finding cosine distance between vectors in this 5000-D space to compute similarity**

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
similarity[1] #similarity of second movie to every other movie
```

<h2>Main Recommender Function</h2>
<hr>

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Spectre')
```
